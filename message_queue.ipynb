```python

from queue import Queue
import time

# 1. Create a message queue
message_queue = Queue()

# 2. Sample order data
orders = [
    {"id": 101, "customer": "Rahul", "product": "Laptop", "amount": 60000},
    {"id": 102, "customer": "Priya", "product": "Mouse", "amount": 800},
    {"id": 103, "customer": "Aman", "product": "Keyboard", "amount": 2000}
]

# 3. Server puts orders into the queue
for order in orders:
    print(f"Server: Order {order['id']} received")
    message_queue.put(order)

print("\nAll orders added to queue.\n")
```

```python
# 4. Worker consumes and processes messages
def worker():
    while not message_queue.empty():

        # Get one message from the queue
        order = message_queue.get()

        print(f"Worker: Processing order {order['id']}")

        # Simulate a task taking time
        time.sleep(1)

        # Simulate successful processing
        print(f"Worker: Order {order['id']} processed")
        print(f"Worker: ACK sent for order {order['id']}\n")

        # Mark this task as completed
        message_queue.task_done()

# 5. Start the worker

worker()

print("All tasks completed!")
```
